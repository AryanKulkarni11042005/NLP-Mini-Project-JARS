```python
from pathlib import Path
import sys
PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "evaluation").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))

import re, random, difflib, os, tempfile, pickle, unicodedata
import numpy as np, pandas as pd
from collections import Counter
from functools import lru_cache
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score
from indic_transliteration import sanscript
from indic_transliteration.sanscript import transliterate
import joblib
import torch, torch.nn as nn
import fasttext

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(0)
dev = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(dev)
```

```python
from datasets import load_dataset

ds = load_dataset("OdiaGenAI/sentiment_analysis_hindi")
df = pd.concat([ds[k].to_pandas() for k in ds.keys()], ignore_index=True)

LABEL_MAP = {"pos": "positive", "neu": "neutral", "neg": "negative"}
d = df[["text", "label"]].rename(columns={"text": "Sentence", "label": "Label"})
d["Label"] = d["Label"].astype(str).str.strip().str.lower().map(LABEL_MAP)
print("unmapped labels:", d["Label"].isna().sum())
d = d.dropna()
print(d["Label"].value_counts())
```

```python
from evaluation.transforms import clean_hindi

d["Sentence"] = d["Sentence"].astype(str)

n0 = len(d)
d = d.drop_duplicates(subset="Sentence")
print("duplicates removed:", n0 - len(d))
d["Sentence"] = d["Sentence"].apply(clean_hindi)
d = d[d["Sentence"].str.len() > 0].reset_index(drop=True)

X_train, X_test, y_train, y_test = train_test_split(
    d["Sentence"], d["Label"], test_size=0.2, random_state=SEED, stratify=d["Label"]
)
print(len(X_train), len(X_test))
```

```python
from evaluation.transforms import to_hinglish

X_train_hing = X_train.apply(to_hinglish)
X_test_hing = X_test.apply(to_hinglish)

print("still has Devanagari:", X_test_hing.str.contains(r"[\u0900-\u097F]").mean())
print("empty rows:", (X_test_hing.str.len() == 0).sum())
print(X_test_hing.iloc[0])
```

```python
from evaluation.transforms import add_noise, fit_normalizer, NOISE_LEVELS, NOISE_SEEDS

normalize, normalizer_state = fit_normalizer(X_train_hing.tolist())
freq = normalizer_state["freq"]
vocab_set = normalizer_state["vocab_set"]
by_len = normalizer_state["by_len"]
vocab = normalizer_state["vocab"]
LEVELS = NOISE_LEVELS
SEEDS = NOISE_SEEDS

NOISY, NORM = {}, {}
for name, level in LEVELS.items():
    for seed in SEEDS:
        NOISY[(name, seed)] = X_test_hing.apply(lambda text: add_noise(text, level, seed))
        NORM[(name, seed)] = NOISY[(name, seed)].apply(normalize)

print(len(NOISY), "noisy sets built,", len(vocab), "words in normalizer vocabulary")
print(NOISY[("High", 1)].iloc[0])
print(NORM[("High", 1)].iloc[0])
```

```python
def mf1(pred_fn, X):
    return f1_score(y_test, pred_fn(X), average="macro", zero_division=0)

def full_eval(pred_fn):
    res = {"Clean Hindi": mf1(pred_fn, X_test), "Clean Hinglish": mf1(pred_fn, X_test_hing)}
    for name in LEVELS:
        res[f"{name} noisy"] = np.mean([mf1(pred_fn, NOISY[(name, s)]) for s in SEEDS])
        res[f"{name} normalized"] = np.mean([mf1(pred_fn, NORM[(name, s)]) for s in SEEDS])
    return res

X_both = pd.concat([X_train, X_train_hing])
y_both = pd.concat([y_train, y_train])
CLASSES = sorted(set(y_both))
deployed = {}
print(CLASSES)
```

```python
try:
    from google.colab import drive
    drive.mount("/content/drive")
    EXPORT = "/content/drive/MyDrive/hindi_sentiment_product"
except ImportError:
    root = os.getcwd()
    if not os.path.isdir(os.path.join(root, "model")) and os.path.isdir(os.path.join(os.path.dirname(root), "model")):
        root = os.path.dirname(root)
    EXPORT = os.path.join(root, "model")
print("Exporting to:", os.path.abspath(EXPORT))
for sub in ["tfidf", "lstm", "fasttext"]:
    os.makedirs(f"{EXPORT}/{sub}", exist_ok=True)

pickle.dump({"vocab_set": vocab_set, "by_len": by_len, "freq": freq, "classes": CLASSES},
            open(f"{EXPORT}/normalizer.pkl", "wb"))
print(os.listdir(EXPORT))
```

```python
tfidf_vec = TfidfVectorizer(token_pattern=r"\S+")
tfidf_lr = LogisticRegression(max_iter=1000, class_weight="balanced").fit(tfidf_vec.fit_transform(X_both), y_both)

def tfidf_pred(T):
    return tfidf_lr.predict(tfidf_vec.transform(T))

deployed["TF-IDF + LogReg"] = full_eval(tfidf_pred)
joblib.dump({"vec": tfidf_vec, "model": tfidf_lr}, f"{EXPORT}/tfidf/tfidf.joblib")
print(pd.DataFrame(deployed).round(3))
```

```python
class LSTMNet(nn.Module):
    def __init__(self, vocab_size, n_classes):
        super().__init__()
        self.emb = nn.Embedding(vocab_size, 100, padding_idx=0)
        self.lstm = nn.LSTM(100, 128, batch_first=True, bidirectional=True)
        self.drop = nn.Dropout(0.4)
        self.fc = nn.Linear(256, n_classes)

    def forward(self, x):
        mask = (x != 0).unsqueeze(-1)
        h, _ = self.lstm(self.emb(x))
        return self.fc(self.drop(h.masked_fill(~mask, -1e9).max(1).values))

def make_encoder(itos, maxlen=40):
    stoi = {w: i for i, w in enumerate(itos)}
    def encode(T):
        out = torch.zeros(len(T), maxlen, dtype=torch.long)
        for i, t in enumerate(T):
            ids = [stoi.get(w, 1) for w in t.split()][:maxlen]
            if ids:
                out[i, :len(ids)] = torch.tensor(ids)
        return out
    return encode

def train_lstm(X, y, epochs=15, maxlen=40):
    texts, labs = X.tolist(), y.tolist()
    classes = sorted(set(labs))
    l2i = {l: i for i, l in enumerate(classes)}
    yt = torch.tensor([l2i[l] for l in labs])
    cnt = Counter(w for s in texts for w in s.split())
    itos = ["<pad>", "<unk>"] + [w for w, c in cnt.items() if c >= 2]
    encode = make_encoder(itos, maxlen)

    net = LSTMNet(len(itos), len(classes)).to(dev)
    Xt = encode(texts)
    w = (len(yt) / (len(classes) * torch.bincount(yt).float())).to(dev)
    loss_fn = nn.CrossEntropyLoss(weight=w)
    opt = torch.optim.Adam(net.parameters(), lr=1e-3)
    for ep in range(epochs):
        net.train()
        perm = torch.randperm(len(yt))
        tot = 0
        for i in range(0, len(perm), 64):
            idx = perm[i:i+64]
            opt.zero_grad()
            loss = loss_fn(net(Xt[idx].to(dev)), yt[idx].to(dev))
            loss.backward()
            opt.step()
            tot += loss.item() * len(idx)
        print(f"epoch {ep+1} loss {tot/len(yt):.3f}")
    return net, itos, classes

torch.manual_seed(0)
lstm_net, lstm_itos, lstm_classes = train_lstm(X_both, y_both)
lstm_encode = make_encoder(lstm_itos)

def lstm_pred(T):
    lstm_net.eval()
    E = lstm_encode(T.tolist())
    preds = []
    with torch.no_grad():
        for i in range(0, len(E), 512):
            preds += lstm_net(E[i:i+512].to(dev)).argmax(1).tolist()
    return [lstm_classes[i] for i in preds]

deployed["LSTM"] = full_eval(lstm_pred)
torch.save({"state_dict": {k: v.cpu() for k, v in lstm_net.state_dict().items()},
            "itos": lstm_itos, "classes": lstm_classes, "maxlen": 40},
           f"{EXPORT}/lstm/lstm.pt")
print(pd.DataFrame(deployed).round(3))
```

```python
def train_fasttext(X, y):
    rows = list(zip(X.tolist(), y.tolist()))
    cnt = Counter(l for _, l in rows)
    top = max(cnt.values())
    extra = []
    for l, c in cnt.items():
        pool = [r for r in rows if r[1] == l]
        extra += [pool[i % len(pool)] for i in range(top - c)]
    rows += extra
    random.Random(0).shuffle(rows)

    path = os.path.join(tempfile.mkdtemp(), "train.txt")
    with open(path, "w", encoding="utf-8") as f:
        for t, l in rows:
            f.write(f"__label__{l} {t}\n")

    return fasttext.train_supervised(path, epoch=25, lr=0.5, wordNgrams=1, minn=2, maxn=5,
                                     dim=50, seed=0, thread=1, verbose=0)

ft_fallback = Counter(y_both).most_common(1)[0][0]
ft_model = train_fasttext(X_both, y_both)

def ft_pred(T):
    out = []
    for t in T:
        p = ft_model.f.predict(t.replace("\n", " ") + "\n", 1, 0.0, "strict")
        out.append(p[0][1].replace("__label__", "") if p else ft_fallback)
    return out

deployed["fastText"] = full_eval(ft_pred)
ft_model.save_model(f"{EXPORT}/fasttext/fasttext.bin")
print(pd.DataFrame(deployed).round(3))
```

```python
tf_saved = joblib.load(f"{EXPORT}/tfidf/tfidf.joblib")
ck = torch.load(f"{EXPORT}/lstm/lstm.pt", map_location="cpu")
lstm_loaded = LSTMNet(len(ck["itos"]), len(ck["classes"]))
lstm_loaded.load_state_dict(ck["state_dict"])
lstm_loaded.eval()
lstm_loaded_encode = make_encoder(ck["itos"], ck["maxlen"])
ft_loaded = fasttext.load_model(f"{EXPORT}/fasttext/fasttext.bin")

def tf_loaded_pred(T):
    return tf_saved["model"].predict(tf_saved["vec"].transform(T))

def lstm_loaded_pred(T):
    with torch.no_grad():
        p = lstm_loaded(lstm_loaded_encode(T.tolist())).argmax(1).tolist()
    return [ck["classes"][i] for i in p]

def ft_loaded_pred(T):
    out = []
    for t in T:
        p = ft_loaded.f.predict(t.replace("\n", " ") + "\n", 1, 0.0, "strict")
        out.append(p[0][1].replace("__label__", "") if p else ft_fallback)
    return out

checks = {
    "TF-IDF + LogReg": (tfidf_pred, tf_loaded_pred),
    "LSTM": (lstm_pred, lstm_loaded_pred),
    "fastText": (ft_pred, ft_loaded_pred),
}
for name, (orig, loaded) in checks.items():
    for label, X in [("Hindi", X_test), ("Hinglish", X_test_hing)]:
        same = np.mean(np.array(list(orig(X))) == np.array(list(loaded(X))))
        print(f"{name} | {label} test | saved model agrees with trained model on {same:.1%}")
```

## Separate text-transformation and classification evaluation

Text metrics compare each noisy/normalized held-out Hinglish candidate with its paired clean Hinglish reference. These references are generated from the same held-out Hindi sentence by the notebook transliterator; there are no independent human Hinglish references, so Hindi-to-Hinglish translation quality is not claimed. Clean/noisy/normalized lexical scores describe preservation or restoration only.

Classification metrics are computed separately from labels. Macro F1 is the primary robustness measure; accuracy, macro precision/recall, and weighted F1 are retained as supporting measures. Five noise seeds are summarized with means and sample standard deviations.

BLEU/ROUGE/METEOR/chrF are higher-is-better lexical overlap scores. TER is reported separately and lower is better. The shared module records tokenizer, dataset, representation, condition, and seed metadata.


```python
import pandas as pd
from evaluation.text_metrics import evaluate_transformation
from evaluation.robustness_metrics import summarize_replicates

# There is no independent human Hinglish reference in this dataset. The paired
# clean reference is the deterministic transliteration of the same held-out Hindi
# sentence. It is suitable for measuring preservation/restoration, not human
# transliteration quality.
text_metric_rows = []
refs = X_test_hing.tolist()
text_metric_rows.append(evaluate_transformation(
    refs, refs, condition="Clean Hinglish", noise_level="none",
    normalization=False, seed="clean", dataset="Hindi product reviews",
))
for level_name in LEVELS:
    for seed in SEEDS:
        text_metric_rows.append(evaluate_transformation(
            refs, NOISY[(level_name, seed)].tolist(),
            condition=f"{level_name} noisy", noise_level=level_name,
            normalization=False, seed=seed, dataset="Hindi product reviews",
        ))
        text_metric_rows.append(evaluate_transformation(
            refs, NORM[(level_name, seed)].tolist(),
            condition=f"{level_name} normalized", noise_level=level_name,
            normalization=True, seed=seed, dataset="Hindi product reviews",
        ))

text_metrics_runs = pd.DataFrame(text_metric_rows)
text_summary = summarize_replicates(
    text_metrics_runs,
    ["condition", "noise_level", "normalization", "dataset", "language", "script"],
    ["bleu", "bleu_1", "bleu_2", "bleu_3", "bleu_4", "rouge1", "rouge2", "rougeL", "meteor", "chrf", "ter"],
)
print(text_summary.to_string(index=False))
```

```python
import numpy as np
import pandas as pd
from evaluation.classification_metrics import evaluate_classification_conditions
from evaluation.robustness_metrics import (
    CLASSIFICATION_METRICS, exploratory_correlations, robustness_table, summarize_replicates,
)
from evaluation.artifacts import write_csv_new
from evaluation.plots import (
    plot_classification_robustness, plot_normalization_recovery,
    plot_text_quality, plot_quality_correlations,
)

# These are separate classification measures. Predictor wrappers preserve the
# existing trained notebook models and do not retrain them in this cell.
predictors = {
    "TF-IDF + LogReg": lambda texts: tfidf_pred(pd.Series(texts)),
    "LSTM": lambda texts: lstm_pred(pd.Series(texts)),
    "fastText": lambda texts: ft_pred(texts),
}
# Include the separately fine-tuned IndicBERT artifact only when it is already
# available in the checkout. Do not download model weights as part of evaluation.
try:
    import sentiment as sentiment_runtime
    if "IndicBERT" in sentiment_runtime.available_models():
        indicbert_raw_predict = sentiment_runtime._get("IndicBERT")
        def predict_indicbert_without_normalization(texts):
            labels = []
            for text in texts:
                probs = indicbert_raw_predict(text)
                labels.append(max(probs, key=probs.get))
            return labels
        predictors["IndicBERT"] = predict_indicbert_without_normalization
except (ImportError, FileNotFoundError):
    pass

conditions = {
    "Clean Hindi": [{"texts": X_test.tolist(), "noise_level": "none", "normalization": False,
                     "seed": "clean", "language": "Hindi", "script": "Devanagari"}],
    "Clean Hinglish": [{"texts": X_test_hing.tolist(), "noise_level": "none", "normalization": False,
                        "seed": "clean", "language": "Hinglish", "script": "Latin"}],
}
for level_name in LEVELS:
    conditions[f"{level_name} noisy"] = []
    conditions[f"{level_name} normalized"] = []
    for seed in SEEDS:
        conditions[f"{level_name} noisy"].append({
            "texts": NOISY[(level_name, seed)].tolist(), "noise_level": level_name,
            "normalization": False, "seed": seed,
            "language": "Hinglish", "script": "Latin",
        })
        conditions[f"{level_name} normalized"].append({
            "texts": NORM[(level_name, seed)].tolist(), "noise_level": level_name,
            "normalization": True, "seed": seed,
            "language": "Hinglish", "script": "Latin",
        })

classification_rows = evaluate_classification_conditions(
    y_test.tolist(), predictors, conditions,
    dataset="Hindi product reviews", language="Hinglish", script="Latin",
)
classification_runs = pd.DataFrame(classification_rows)
classification_summary = summarize_replicates(
    classification_runs,
    ["model", "dataset", "language", "script", "condition", "noise_level", "normalization"],
    list(CLASSIFICATION_METRICS),
)
robustness = robustness_table(classification_runs)
print("Classification metrics by model and condition (mean and sample SD):")
print(classification_summary.to_string(index=False))
print("\nMacro-F1 drop and high-noise recovery:")
print(robustness.to_string(index=False))

correlations, paired_quality_f1 = exploratory_correlations(text_metrics_runs, classification_runs)
print("\nExploratory quality/F1 correlations (coefficients only; not inferential):")
print(correlations.to_string(index=False))

output_dir = Path(EXPORT) / "evaluation_outputs"
output_dir.mkdir(parents=True, exist_ok=True)
outputs = [
    write_csv_new(text_metrics_runs, output_dir / "results_text_metrics_runs.csv"),
    write_csv_new(text_summary, output_dir / "results_text_metrics_summary.csv"),
    write_csv_new(classification_runs, output_dir / "results_classification_runs.csv"),
    write_csv_new(classification_summary, output_dir / "results_classification_summary.csv"),
    write_csv_new(robustness, output_dir / "results_classification_robustness.csv"),
    write_csv_new(correlations, output_dir / "results_quality_f1_correlations.csv"),
    write_csv_new(paired_quality_f1, output_dir / "results_quality_f1_pairs.csv"),
]
plot_dir = output_dir / "plots"
plots = [plot_classification_robustness(classification_summary, plot_dir)]
plots.append(plot_normalization_recovery(classification_runs, plot_dir))
plots.extend(plot_text_quality(text_summary, plot_dir))
plots.extend(plot_quality_correlations(paired_quality_f1, plot_dir))
plots = [path for path in plots if path is not None]
print("\nSaved result files:")
print("\n".join(str(path) for path in outputs))
print("Saved plots:")
print("\n".join(str(path) for path in plots))
```
